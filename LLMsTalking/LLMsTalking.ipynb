```python
import ollama
from openai import OpenAI
import os
from dotenv import load_dotenv
from IPython.display import display, Markdown
```

```python
GPT_MODEL = "gpt-4o-mini"
OLLAMA_MODE = "llama3.2"
```

```python
load_dotenv()
key = os.getenv("OPENAI_API_KEY")
```

```python
openai = OpenAI()
```

```python
gpt_system = "You are a stubborn chatbot that loves action movies and hate drama movies and think they are terrible. You always respond in arrogance"
ollama_system = "You are a stubborn chatbot that loves drama movies and hate action movies and think they are terrible. You always respond in arrogance"
```

```python
start_msg = "Hi"
```

```python
gpt_messages = [
    {"role": "system", "content": gpt_system},
    {"role": "user", "content": start_msg}
]
```

```python
ollama_messages = [
    {"role": "system", "content": ollama_system},
    {"role": "assistant", "content": start_msg}
]
```

```python
def gen_gpt():
    response = openai.chat.completions.create(
        model=GPT_MODEL,
        messages=gpt_messages
    )

    return response.choices[0].message.content
```

```python
def gen_ollama():
    response = ollama.chat(
        model=OLLAMA_MODE,
        messages=ollama_messages
    )

    return response["message"]["content"]
```

```python
def explode(conversations: int):
    for i in range(conversations):
        gpt_response = gen_gpt()
        gpt_messages.append(
            {"role": "assistant", "content": gpt_response}
        )
        ollama_messages.append(
            {"role": "user", "content": gpt_response}
        )

        display(Markdown(f"**GPT**: {gpt_response}\n\n"))

        ollama_response = gen_ollama()
        gpt_messages.append(
            {"role": "user", "content": ollama_response}
        )
        ollama_messages.append(
            {"role": "assistant", "content": ollama_response}
        )

        display(Markdown(f"**OLLAMA**: {ollama_response}\n\n"))
```

```python
explode(5)
```
